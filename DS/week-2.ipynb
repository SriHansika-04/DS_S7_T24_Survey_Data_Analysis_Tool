```python
#handling missing values
#1. mean/median/mode imputation
import pandas as pd
import numpy as np

#sample dataset
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
                'Department':['HR','Finance',np.nan,'IT','null']})

#display original data set
print("Original data set with the missing values")
print(df)
print("\n")

#mean
df['Age']=df['Age'].fillna(df['Age'].mean())
df['Department']=df['Department'].fillna(df['Department'].mode()[0])
print(df)
```

```python
#forward/backward fill
import pandas as pd
import numpy as np
#sample dataset
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
                'Department':['HR','Finance',np.nan,'IT','null']})

#display original data set
print("Original data set with the missing values")
print(df)
print("\n")

#forward fill(use previous value)
df_ffill=df.copy()
df_ffill.ffill(inplace=True)
print(df_ffill)

#backwardfill(use next value)
df_bfill=df.copy()
df_bfill.bfill(inplace=True)
print(df_bfill)
```

```python
#ELETION-DROP ROW OR column 
import pandas as pd
import numpy as np
#sample dataset
df=pd.DataFrame({'Age':[25,30,np.nan,40,35],
                'Department':['HR','Finance','IT','Finance','IT']})

#display original data set
print("Original data set with the missing values")
print(df)
print("\n")
df_drop_rows=df.dropna()
print("After dropping rows:\n",df_drop_rows)

df_drop_col=df.dropna()
print("After dropping columns:\n",df_drop_col)
```

```python
#removing duplicates 
import pandas as pd
df=pd.DataFrame({
    'ID':[1,2,2,3,4,4],
    'Name':['Alice','Bob','Bob','Charlie','David','David'],
    'Age':[25,30,30,35,40,40]
})
print("Original data: \n",df)

#remove duplicates
df_exact=df.drop_duplicates()
print("\n After Exact match removal: \n",df_exact)
```

```python
#removing duplicates-subset based removal
import pandas as pd
df=pd.DataFrame({
    'ID':[1,2,2,3,4,4],
    'Name':['Alice','Bob','Bob','Charlie','David','David'],
    'Age':[25,30,30,35,40,40]
})
print("Original data: \n",df)

df_subset_id=df.drop_duplicates(subset=['ID'])
print("\n after the subset based removal (id):\n",df_subset_id)

df_subset_name=df.drop_duplicates(subset=['Name'])
print("\n after the subset based removal (name):\n",df_subset_name)
```

```python
#removing duplicates-correcting inconsistent format
import pandas as pd
df=pd.DataFrame({
    'Date':['2025-01-05','05/01/2025','Jan 5,2025','2025.01.05']
})
print("Original data: \n",df)
print("\n")
#convert all dates to standard ISO format(YYYY-MM-DD)
df['Date']=pd.to_datetime(df['Date'],errors='coerce').dt.strftime('%Y-%m-%d')
print(df)
```

```python
#Removing  Duplicates(Correcting Inconsistent Format)
#3.b.b Case Normalization
df=pd. DataFrame({
    'Name':['MOUNI','mouni','Mounitha','MoUnI']
})
print("Original Data:\n",df)

df['Name_lower']=df['Name'].str.lower()

df['Name_upper']=df['Name'].str.upper()

print("\n",df)
```
