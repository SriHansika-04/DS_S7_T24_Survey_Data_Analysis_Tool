```python
#similarity and dissmilarity measures for data analytics
import numpy as np
from scipy.spatial import distance

pointA=np.array([2,4,6])
pointB=np.array([5,1,9])

#euclidean distance-find how far these two points A and B are from each other
eculidean_dist=distance.euclidean(pointA,pointB)
print("Euclidean distance:",eculidean_dist)

#similarity-inverse of distance
similarity_dist= 1/(1+eculidean_dist)
print("Euclidean similarity:",similarity_dist)
```

```python
#vector based distance measures(manhattan distance)
import numpy as np
from scipy.spatial import distance
pointA=np.array([2,4,6])
pointB=np.array([5,1,9])
#manhattan distance 
Manhattan_dist=distance.cityblock(pointA,pointB)
print("Manhattan distance:",Manhattan_dist)

#similarity-inverse of distance
sim_Manhattan=1/(1+Manhattan_dist)
print("Manhattan similarity:",sim_Manhattan)
```

```python
#vector-based distance measures (minkowski distance)

import numpy as np
from scipy.spatial import distance

pointA=np.array([2,4,6])
pointB=np.array([5,1,9])

#Minkowski distance with p=3
minkowski_dist_p3=distance.minkowski(pointA,pointB,p=3)
print("Minkowski Distance(p=3):",minkowski_dist_p3)

#similarity
sim_minkowski=1/(1+minkowski_dist_p3)
print("Minkowski Similarity (p=3)",sim_minkowski)
```

```python
#spearmen correlation
import pandas as pd
from scipy.stats import spearmanr

df=pd.DataFrame({
    'X':[10,20,30,40,50],
    'Y':[12,18,33,47,55]
})
corr_val , p_val = spearmanr(df['X'],df['Y'])
print(f"Spearman Correlation Coefficient: {corr_val}")
print(f"P-value: {p_val}")
```

```python
#edit-based measures (string/text data-hamming distance )
def hamming_dist(str1,str2):
    #ensure lengths are equal
    if len(str1)!= len(str2):
        raise Valueerror("Strings must be of equal length")

    return sum(ch1 != ch2 for ch1, ch2 in zip(str1,str2))

s1="karolin"
s2="kathrin"

dist=hamming_dist(s1,s2)
print(f"Hamming Distance between '{s1}' and '{s2}':{dist}")
    


```

```python
#jaccard indexx-compare text doc using the words present in them

def jaccard_index(str1,str2):
    set1,set2=set(str1.split()),set(str2.split())
    intersection=set1.intersection(set2)
    union=set1.union(set2)
    return len(intersection)/len(union)

s1="data science is fun"
s2=" science makes data useful"

print("jaccard Index: ",jaccard_index(s1,s2))

```
