```python
#slicing string
""" syntax=[start:stop:step]  
start is starting index and stop is ending index 
step is the no of postions to move (optional)"""
#start index is included and stop index is excluded
b= "KLH BOWRAMPET"
print(b[2:5])

```

```python
#types of slices
#slice from front
b= "KLH BOWRAMPET"
#didnt mention any starting index so it will start from 0 default from the beginning till 7 as 8 is stop index
print(b[:8])
```

```python
#slicing from stop
# didnt mention any stop index only start index so slicing starts from the start index mentioned and does till the ending of the string defaultly
b= "KLH BOWRAMPET"
print(b[2:])
```

```python
#negative indexing-we use it so start the slice from the end of the string & the -ve index starts from the end character
b="KONERU LAKSHMAIAHA EDUCATIONAL FOUNDATION"
print(b[-10:-4])
```
