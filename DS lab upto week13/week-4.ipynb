```python
import matplotlib.pyplot as plt
y1=[] #stores +ve values
y2=[] # store -ve value
x=range(-100,100) #generate values from -100 to 90 with an intevral of 10(-100,-90,-80,-70...)
for i in x: 
    y1.append(i**2) #y=x^2
for i in x:
    y2.append(-i**2)

plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000,2000) #y axis limit
plt.axhline(0) #horixantal
plt.axvline(0) # vertical

plt.savefig("quad.png")
plt.show()



    
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris

iris = load_iris()

df = pd.DataFrame(iris.data, columns=iris.feature_names)

plt.subplot(2,3,1)
plt.plot(df['sepal length (cm)'], df['petal length (cm)'])

plt.subplot(2,3,2)
plt.plot(df['sepal width (cm)'], df['petal width (cm)'])

plt.subplot(2,3,3)
plt.scatter(df['sepal length (cm)'],
            df['petal length (cm)'],
            c='green')

plt.title('subplots')
plt.show()
```
